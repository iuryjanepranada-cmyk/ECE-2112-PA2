# **Experiment #2 - Numerical Python (NUMPY)**

```python
import numpy as np
```

### **A. REPRODUCIBLE NORMALIZATION PROBLEM**

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
X_normalized = (X - X.mean())/(X.std())
X_normalized
```

```python
X_normalized.mean()
```

```python
X_normalized.std()
```

```python
np.save("X_normalized.npy", X_normalized)
```

### **B. CUBES DIVISIBLE BY 4 PROBLEMB. CUBES DIVISIBLE BY 4 PROBLEM**

```python
x = np.arange(1,101)
x
```

```python
C = x**3
C
```

```python
C.reshape(10,10).shape
```

```python
C.resize(10,10)
C
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
np.save("div_by_4.npy", div_by_4)
```

### **C. ABOVE-MEAN SQUARES PROBLEM**

```python
x = np.arange(1,37) ** 2
x
```

```python
S = x
S = np.resize(S, (6,6))
S
```

```python
S_mean = S.mean()
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
above_mean.size
```

```python
np.save("above_mean.npy", above_mean)
```
